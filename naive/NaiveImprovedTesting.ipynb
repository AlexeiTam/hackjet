```python
import naive
```

```python
InputJets = naive.JetReader("PythiaEventsBatchTest.root",'pdEventTree')
```

```python
OutHistory_kT, OutJets_kT = naive.gen_clustering(1, InputJets,0.4)
OutHistory_CA, OutJets_CA = naive.gen_clustering(0, InputJets,0.4)
OutHistory_antikT, OutJets_antikT = naive.gen_clustering(-1, InputJets,0.4)
```

```python
for jet in OutJets_CA:
  print(jet)
```

```python
naive.JetWriter("naive_antikt.root",OutJets_antikT,"finalJets")
naive.JetWriter("naive_kt.root",OutJets_kT,"finalJets")
naive.JetWriter("naive_ca.root",OutJets_CA,"finalJets")
```
