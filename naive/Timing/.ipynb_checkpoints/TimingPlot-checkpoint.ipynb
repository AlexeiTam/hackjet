```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
naive_data = pd.read_csv('timing_data.csv')
```

```python
naive_data
```

```python
fj_data = pd.read_csv('fj_timing_data.csv')

fj_data
```

```python
t_naive = naive_data["t"]
dt_naive = naive_data["dt"]
N_naive = naive_data["N"]

t_fj = fj_data["t"]
dt_fj = fj_data["dt"]
N_fj = fj_data["N"]
```

```python
plt.errorbar(N_naive,t_naive,yerr=dt_naive,fmt="o",label="Naive")
plt.errorbar(N_fj,t_fj,yerr=dt_fj,fmt="o",label="FJ")
plt.show()
```

```python
plt.errorbar(N_naive,t_naive,yerr=dt_naive,label="Naive")
plt.errorbar(N_fj,t_fj,yerr=dt_fj,label="FJ")
plt.show()
```
