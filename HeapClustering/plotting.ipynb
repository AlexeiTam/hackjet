```python
import matplotlib.pyplot as plt
import pandas as pd


tvN_prefix = pd.read_csv('timing_heap_preWhere.csv')
tvN_heap = pd.read_csv('timing_heap.csv')
tvN_naive = pd.read_csv('timing_naive.csv')


tp_data = tvN_prefix["t"]
dtp_data = tvN_prefix["dt"]
Np_data = tvN_prefix["N"]

t_data = tvN_heap["t"]
dt_data = tvN_heap["dt"]
N_data = tvN_heap["N"]

tn_data = tvN_naive["t"]
dtn_data = tvN_naive["dt"]
Nn_data = tvN_naive["N"]



plt.errorbar(N_data, t_data, yerr=dt_data, fmt="o", label="heap")
plt.errorbar(Nn_data, tn_data, yerr=dtn_data, fmt="o", label="naive")
plt.errorbar(Np_data, tp_data, yerr=dtp_data, fmt="o", label="heap (*)")
plt.xlabel("N")
plt.ylabel("t (s)")
plt.legend()
plt.show()
```
